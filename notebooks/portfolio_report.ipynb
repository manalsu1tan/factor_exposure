# Portfolio Report Notebook

Lightweight report for:
- Views expressed
- Top risk contributors
- Exposure drift over time

```python
from datetime import date
import polars as pl

from factor_exposure.model.artifacts import load_latest_artifacts
from factor_exposure.reporting.report import build_portfolio_report
```

```python
holdings = [
    ("AAPL", 0.5),
    ("MSFT", 0.5),
]

as_of = date(2025, 12, 31)
start_date = date(2025, 1, 1)
end_date = date(2025, 12, 31)

artifacts = load_latest_artifacts()
report = build_portfolio_report(
    holdings=holdings,
    artifacts=artifacts,
    as_of=as_of,
    start_date=start_date,
    end_date=end_date,
    top_n=5,
)
```
